```python
import psycopg2
import re

conn = psycopg2.connect(
    dbname="postgres",
    user="postgres",
    password="postgres",
    host="localhost",  # e.g., "localhost" or an IP address
    port="5432"  # Default PostgreSQL port
)

cur = conn.cursor()

# Example query
cur.execute("""
SELECT title, url, DATE(published_time) FROM news_urls
WHERE
    title IS NOT NULL AND
    title LIKE '%Tạp Chí%' AND
    published_time IS NOT NULL AND
    published_time >= '2025-03-07'
""")
rows = cur.fetchall()

# Close the cursor and connection


def clean_news_title(title: str) -> str:
    return re.sub(r"\s*-\s*Tạp Chí.*", "", title)

def add_back(title: str, url: str):
    cleaned_title = clean_news_title(title)
    print(cleaned_title)
    insert_query = """INSERT INTO news_urls (url, title) VALUES (%s, %s)
        ON CONFLICT (url) DO UPDATE SET
        title = EXCLUDED.title
        """
    cur.execute(insert_query, (url, cleaned_title))
    conn.commit()

for r in rows:
    add_back(r[0], r[1])

cur.close()
conn.close()
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def generate_synthetic_data():
    np.random.seed(42)
    base_rate = 10  # Số lần xuất hiện trung bình mỗi ngày
    data = np.random.poisson(lam=base_rate, size=7)  # Dữ liệu 7 ngày
    
    # Giả lập trend vào ngày 11-13/3
    data[4:7] += np.random.randint(5, 15, size=3)  
    return [0, 1, 2, 1, 2, 2, 2]

def detect_poisson_trend(data):
    trends = []
    for i in range(len(data)):
        lambda_est = np.mean(data[:i]) if i > 0 else np.mean(data)
        threshold = stats.poisson.ppf(0.99, lambda_est)  # Ngưỡng 99%
        is_trend = data[i] > threshold
        trends.append(is_trend)
    return trends

def detect_z_score_trend(data):
    trends = []
    for i in range(len(data)):
        if i == 0:
            trends.append(False)
            continue
        mean = np.mean(data[:i])
        std = np.std(data[:i]) if np.std(data[:i]) > 0 else 1  # Tránh chia cho 0
        z_score = (data[i] - mean) / std
        is_trend = z_score > 2
        trends.append(is_trend)
    return trends

def detect_cusum_trend(data):
    k = 0.5  # Giá trị tham chiếu
    h = 5    # Ngưỡng phát hiện trend
    cusum = np.zeros(len(data))
    trends = [False] * len(data)
    
    for i in range(1, len(data)):
        cusum[i] = max(0, cusum[i-1] + (data[i] - np.mean(data[:i]) - k))
        if cusum[i] > h:
            trends[i] = True
    
    return trends, cusum

# Tạo dữ liệu
data = generate_synthetic_data()
dates = ["7/3", "8/3", "9/3", "10/3", "11/3", "12/3", "13/3"]

# Áp dụng các phương pháp
poisson_trends = detect_poisson_trend(data)
z_score_trends = detect_z_score_trend(data)
cusum_trends, cusum_values = detect_cusum_trend(data)

# Hiển thị kết quả
print("Ngày\t\tSố lần xuất hiện\tPoisson\tZ-score\tCUSUM")
for i in range(len(data)):
    print(f"{dates[i]}\t{data[i]}\t\t{'✔' if poisson_trends[i] else '✘'}\t"
          f"{'✔' if z_score_trends[i] else '✘'}\t"
          f"{'✔' if cusum_trends[i] else '✘'}")

# Vẽ biểu đồ
plt.figure(figsize=(12, 6))
plt.plot(dates, data, marker='o', label='Số lần xuất hiện')
plt.xticks(rotation=45)
plt.xlabel('Ngày')
plt.ylabel('Tần suất')
plt.title('Phát hiện Trend theo ngày bằng Poisson, Z-score và CUSUM')

# Đánh dấu ngày có trend
for i in range(len(data)):
    if poisson_trends[i] or z_score_trends[i] or cusum_trends[i]:
        plt.scatter(dates[i], data[i], color='red', s=100, label='Trend' if i == 0 else "")

plt.legend()
plt.show()
```
